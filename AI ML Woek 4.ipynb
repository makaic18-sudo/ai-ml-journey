```python

import pandas as pd
import numpy as np
data = {
    "name": ["Ali", "Fatima", "Hamza", "Junaid"],
    "age": [24, np.nan, 55, 66],
    "score": [44, 66, np.nan, 77]
}
df = pd.DataFrame(data)
print(df)

print(df.isnull())

print(df.isnull().sum())

print(df.fillna(0))

print(df["score"].fillna(df["score"].mean()))

print(df.dropna())

age_filled = df["age"].fillna(df["age"].mean())
print(age_filled)

print(df.isnull().sum())
```
