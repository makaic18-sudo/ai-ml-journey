```python
import numpy as np
a = np.array([1,2,3,4,5])
print("Array of the whole a:", a)
print("Array of the whole a multiply by 2:",a*2)
print("Array of the whole adds 22:",a+22)
print("Array of the whole a getting its mean:",a.mean())
print("Array of the whole in total:",a.shape)
ali = np.array([4,6,8,22,123,434])
print("Array of the whole ali:",ali.max())
print("Array of the whole ali in total:",ali.sum())
print("Numbers of values in ali:", ali)
```
