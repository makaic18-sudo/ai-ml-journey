```python

# Step 1: Create the CSV file
csv_text = """name,age,score,city
Ali,24,65,Karachi
Fatima,24,66,Lahore
Ayesha,55,77,Karachi
Babar,32,87,Lahore
Virat,38,98,Karachi
"""

with open("students.csv", "w") as f:
    f.write(csv_text)

print("students.csv created successfully!")


# Step 2: Load and explore the CSV file
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("students.csv")

print("\n--- First 5 Rows ---")
print(df.head())

print("\n--- Shape ---")
print(df.shape)

print("\n--- Columns ---")
print(df.columns)

print("\n--- Statistics ---")
print(df.describe())

print("\n--- Last 3 Rows ---")
print(df.tail(3))

print("\n--- Students from Lahore ---")
print(df[df["city"] == "Lahore"])


# Step 3: Bar chart of student scores
plt.bar(df["name"], df["score"])
plt.title("Student Scores")
plt.xlabel("Name")
plt.ylabel("Score")
plt.show()


# Step 4: Bar chart of student ages
plt.bar(df["name"], df["age"])
plt.title("Student Ages")
plt.xlabel("Name")
plt.ylabel("Age")
plt.show()


# Step 5: Line chart of student scores
plt.plot(df["name"], df["score"], marker="o")
plt.title("Student Scores - Line Chart")
plt.xlabel("Name")
plt.ylabel("Score")
plt.show()
```
